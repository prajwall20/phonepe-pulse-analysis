# 📊 Project Summary: PhonePe Pulse Data Analysis

The PhonePe Pulse Data Analysis project focuses on exploring and interpreting large-scale digital payment data to derive meaningful insights into transaction trends, user behavior, and financial product adoption across India. With the rapid growth of digital payment platforms, understanding these patterns is crucial for businesses to make data-driven decisions, optimize services, and enhance user engagement.

This project utilizes the publicly available PhonePe Pulse dataset, which contains structured information on transactions, users, and insurance across different states, districts, and time periods. The data is available in JSON format and is organized into three major categories: Aggregated, Top, and Map data. An end-to-end ETL (Extract, Transform, Load) pipeline was implemented to convert raw JSON files into structured CSV datasets using Python. These datasets were then stored in a MySQL database using phpMyAdmin, enabling efficient querying and analysis.

The analytical phase of the project was carried out using SQL and Python. SQL queries were used to extract key metrics such as total transaction amounts, top-performing states and districts, transaction types, and user engagement levels. These results were then imported into a Jupyter Notebook using Python, where further analysis and visualizations were performed using libraries like Pandas and Matplotlib. The notebook was structured following a standard EDA (Exploratory Data Analysis) format, including queries, outputs, visualizations, and detailed insights.

Key findings from the analysis revealed that certain states such as Maharashtra, Karnataka, and Tamil Nadu consistently lead in digital transaction volumes, indicating strong digital infrastructure and user adoption. Peer-to-peer transactions were identified as the most dominant transaction type, reflecting the widespread use of UPI-based payments. Additionally, mobile device analysis showed that brands like Xiaomi and Samsung have a significant share of the user base, highlighting the dominance of Android devices in the ecosystem. Insurance data indicated a gradual increase in adoption, suggesting growing awareness of financial protection products.

The project also highlights several business use cases, including customer segmentation, market expansion opportunities, and product optimization strategies. For instance, high-performing regions can be targeted for premium services, while low-performing areas can be explored for growth potential. User engagement metrics such as app opens provide valuable insights into customer behavior and retention.

Overall, this project demonstrates strong proficiency in data extraction, database management, SQL querying, and data visualization. It showcases the practical application of data analytics in the fintech domain and provides actionable insights that can support strategic decision-making.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

```python
df = pd.read_sql("""
SELECT state, SUM(amount) AS total_amount
FROM aggregated_transaction
GROUP BY state
ORDER BY total_amount DESC
LIMIT 10;
""", conn)

df.plot(kind='bar', x='state', y='total_amount')
plt.title("Top 10 States by Transaction Amount")
plt.xticks(rotation=45)
plt.show()
```

### 🔍 Why this chart?
A bar chart is used to compare transaction amounts across different states, making it easy to identify top-performing regions.

### 📊 Insights:
- Maharashtra, Karnataka, and Tamil Nadu lead in transaction volume.
- These states have strong digital infrastructure and high user adoption.

### 💼 Business Impact:
- **Positive:** Businesses can target these states for premium services and investments.
- **Negative:** Lower-performing states indicate untapped or weak markets that need awareness and infrastructure development.

```python
df = pd.read_sql("""
SELECT year, SUM(amount) AS total
FROM aggregated_transaction
GROUP BY year;
""", conn)

df.plot(x='year', y='total')
plt.title("Year-wise Transaction Growth")
plt.show()
```

### 🔍 Why this chart?
A line chart is ideal for analyzing trends over time.

### 📊 Insights:
- Significant growth observed after 2020.
- Indicates rapid adoption of digital payments post-pandemic.

### 💼 Business Impact:
- **Positive:** Growing market size provides expansion opportunities.
- **Negative:** Any fluctuations or dips may indicate system or trust issues.

```python
df = pd.read_sql("""
SELECT type, SUM(amount) AS total
FROM aggregated_transaction
GROUP BY type;
""", conn)

df.plot(kind='bar', x='type', y='total')
plt.xticks(rotation=45)
plt.show()
```

### 🔍 Why this chart?
Bar chart helps compare different transaction categories.

### 📊 Insights:

- Recharge and bill payments also contribute significantly.

### 💼 Business Impact:
- **Positive:** Strong P2P usage shows user trust in digital payments.

```python
df = pd.read_sql("""
SELECT name, SUM(amount) AS total
FROM top_transaction
WHERE type='district'
GROUP BY name
ORDER BY total DESC
LIMIT 10;
""", conn)

df.plot(kind='bar', x='name', y='total')
plt.title("Top Districts by Transaction Amount")
plt.xticks(rotation=45)
plt.show()
```

### 🔍 Why this chart?
Bar chart helps compare transaction performance across districts.

### 📊 Insights:
- Metro districts dominate transaction values.
- Urban areas contribute the highest share.

### 💼 Business Impact:
- **Positive:** Focus marketing and partnerships in high-performing districts.
- **Negative:** Rural districts show lower adoption, indicating growth gaps.

```python
df = pd.read_sql("""
SELECT name, SUM(amount) AS total
FROM top_transaction
WHERE type='pincode'
GROUP BY name
ORDER BY total DESC
LIMIT 10;
""", conn)

df.plot(kind='bar', x='name', y='total')
plt.title("Top Pincodes by Transactions")
plt.show()
```

### 🔍 Why this chart?
Bar chart is useful for comparing transaction concentration at pincode level.

### 💼 Business Impact:

```python
df = pd.read_sql("""
SELECT state, SUM(app_opens) AS total
FROM map_user
GROUP BY state
ORDER BY total DESC
LIMIT 10;
""", conn)

df.plot(kind='bar', x='state', y='total')
plt.title("Top States by App Opens")
plt.xticks(rotation=45)
plt.show()
```
